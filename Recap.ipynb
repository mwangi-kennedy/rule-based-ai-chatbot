Numpy on excel.

```python
import numpy as np
import pandas as pd
df = pd.read_excel("Movies_data_2000.xlsx")
df.head()
```

```python
df.isnull().sum()
```

```python
df["Runtime"] = df["Runtime"].fillna(df["Runtime"].mode()[0])
```

```python
df.isnull().sum()
```

Creating Arrays

```python
budgets = df["Budget"].to_numpy()
print(budgets)
```

```python
total_budget = np.sum(budgets)
average_budget = np.mean(budgets)

print("Total budget:", total_budget)
print("Average budget:", average_budget)
```

```python
high_budget_movies = budgets[budgets > 50000000]
print(high_budget_movies)
```

```python
max_budget = np.max(budgets)
min_budget = np.min(budgets)

print("Highest budget:", max_budget)
print("Lowest budget:", min_budget)
```

```python
budgets_inflated = budgets * 1.10
print(budgets_inflated)
```

```python
np.random.seed(42)
random_budgets = np.random.randint(1000000, 250000000, size=5)
print(random_budgets)
```

```python
import pandas as pd
df = pd.read_excel("Movies_data_2000.xlsx")
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

titles = df["Title"].astype(str)
budgets = df["Budget"]

plt.bar(titles, budgets, color="red")
plt.title("Movie Budgets")
plt.xlabel("Movie Title")
plt.ylabel("Budget")
plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
df = pd.read_excel("Movies_data_2000.xlsx")
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.barh(df["Title"].astype(str), df["Revenue"], color="lightgreen")
plt.title("Revenue per Movie")
plt.xlabel("Revenue")
plt.ylabel("Movie Title")
plt.show()
```

```python
top5 = df.sort_values(by="Budget", ascending=False).head(5)

plt.figure(figsize=(7,7))
plt.pie(top5["Budget"], labels=top5["Title"], autopct='%1.1f%%', startangle=140)
plt.title("Top 5 Movie Budgets Share")
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["Budget"], df["Popularity"], marker='o', linestyle='-', color='purple')
plt.title("Movie Popularity")
plt.xlabel("Movie Title")
plt.ylabel("Popularity")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Budget"], bins=5, color="orange", edgecolor="black")
plt.title("Budget Distribution")
plt.xlabel("Budget")
plt.ylabel("Number of Movies")
plt.show()
```
