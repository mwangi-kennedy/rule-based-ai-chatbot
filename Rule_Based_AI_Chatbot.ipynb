# Rule-Based AI Chatbot

This notebook implements a foundational **Rule-Based AI Chatbot** built entirely in Python. Unlike modern generative AI systems that rely on large language models (LLMs) or probabilistic neural networks, this chatbot operates on a deterministic, control-flow paradigm. It leverages exact string matching to map specific user inputs directly to predefined hardcoded responses.

---

###  Architectural Overview
The system relies on an infinite evaluation loop (`while True`) that continuously captures user input via standard I/O streams. The input is normalized to lowercase to ensure basic case-insensitivity before passing through a conditional matrix (`if-elif-else` chain).



---

###  Capabilities Matrix

| Advantages | Disadvantages |
| :--- | :--- |
| **Predictable & Safe:** Zero risk of "hallucinations" or generating inappropriate content; responses are 100% controlled. | **Extreme Fragility:** Incapable of handling typos, synonyms, or variations in phrasing (e.g., "hello!" or "hi there" will fail if not explicitly mapped). |
| **Highly Efficient:** Requires negligible computational overhead (O(1) to O(N) time complexity) and runs perfectly without a GPU. | **No Context Retention:** Operates statelessly; it cannot remember previous turns in the conversation. |
| **Zero Dependencies:** Relies purely on built-in Python standard primitives without needing external ML libraries. | **Scalability Bottleneck:** As the codebase grows, maintaining a massive nested conditional structure becomes messy and inefficient. |

---


```python
# ============================================
# RULE-BASED AI CHATBOT
# ============================================

print("AI Chatbot Started")
print("Type 'bye' anytime to exit.\n")

# Infinite loop to keep chatbot running
while True:

    # Get user input
    user_input = input("You: ").lower()

    # Greeting responses
    if user_input == "hello":
        print("Bot: Hello! Nice to meet you.")

    elif user_input == "hi":
        print("Bot: Hi there!")

    elif user_input == "hey":
        print("Bot: Hey! How can I help you?")

    # Asking about the bot
    elif user_input == "how are you":
        print("Bot: I'm functioning perfectly.")

    elif user_input == "what is your name":
        print("Bot: I am a Rule-Based AI Chatbot.")

    # Simple knowledge responses
    elif user_input == "what is ai":
        print("Bot: AI stands for Artificial Intelligence.")

    elif user_input == "who created you":
        print("Bot: Boy, i was created using Python by  guru.")

    # Time-related response
    elif user_input == "good morning":
        print("Bot: Good morning, Hope you have a productive day.")

    elif user_input == "good night":
        print("Bot: Good night, Rest well.")

    # Exit condition
    elif user_input == "bye":
        print("Bot: Goodbye. Have a great day.")
        break

    # Unknown input
    else:
        print("Bot: Sorry, I don't understand that yet.")
```
